# Outlier Detection

```python
sns.boxplot(df["Fat"])
```

```python
Q1 = df["Fat"].quantile(0.25)
Q3 = df["Fat"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

fat_outliers = df[
    (df["Fat"] < lower) |
    (df["Fat"] > upper)
]

print("Fat outliers:", len(fat_outliers))
```

# Feature Engineering

```python
df["Protein_per_Calorie"] = np.where(
    df["Caloric Value"] > 0,
    df["Protein"] / df["Caloric Value"],
    np.nan
)
```
